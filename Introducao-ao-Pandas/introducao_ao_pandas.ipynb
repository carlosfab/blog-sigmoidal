# Introdução ao Pandas para Análise Exploratória

Material de apoio do artigo [Introdução ao Pandas para Análise Exploratória](https://sigmoidal.ai) no blog da Sigmoidal.

A ideia aqui é **prática**: você vai carregar um arquivo `csv` real, entender o que tem dentro dele, calcular estatísticas, criar um `DataFrame` do zero e tratar valores ausentes. Tudo o que um cientista de dados faz nos primeiros minutos de qualquer projeto.

> O notebook foi pensado para rodar de ponta a ponta no Google Colab gratuito, sem dependências além de `pandas` e `numpy`.


## 1. Importando as bibliotecas

Pandas costuma vir junto com qualquer ambiente Python científico (Anaconda, Colab, Kaggle). Se for instalar do zero na sua máquina, basta `pip install pandas`.

Por convenção, importa-se com o apelido `pd`. Vamos importar também o `numpy`, que será usado mais à frente.


```python
import pandas as pd
import numpy as np

print(f"pandas: {pd.__version__}")
print(f"numpy:  {np.__version__}")
```

## 2. Carregando dados de um arquivo CSV

A função `pd.read_csv()` aceita tanto um caminho local quanto uma URL. Aqui vamos carregar o histórico de cotações da ação ordinária do **Banco do Brasil (BBAS3)**, hospedado num repositório público no GitHub.


```python
url_bbas3 = "https://raw.githubusercontent.com/carlosfab/dsnp2/master/datasets/BBAS3.SA.csv"
df = pd.read_csv(url_bbas3)
df.head()
```

Pandas reconhece muitos outros formatos: JSON, Parquet, HDF, Excel, SQL e por aí vai. A lógica é sempre a mesma, uma função `pd.read_*()` devolvendo um `DataFrame`.

## 4. DataFrame vs Series

Um `DataFrame` é a tabela inteira. Uma `Series` é uma coluna isolada. Dá para selecionar uma coluna por colchetes ou por ponto.


```python
# selecionar uma coluna por colchetes
df["High"].head()
```

```python
# mesma coluna, sintaxe alternativa
df.High.head()
```

```python
# tipo de uma coluna isolada
type(df["High"])
```

## 5. Trabalhando com datas

Repare em `df.dtypes` que a coluna `Date` veio como `object`, ou seja, texto. Pandas não reconhece datas automaticamente, então é preciso converter explicitamente com `pd.to_datetime()`.


```python
df.Date.head()
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%Y-%m-%d")
df.dtypes
```

Com a coluna no tipo `datetime64`, ganhamos o acessor `.dt`, que extrai componentes da data sem complicação.


```python
df["Year"] = df.Date.dt.year
df["Month"] = df.Date.dt.month
df.head()
```

## 6. Estatísticas descritivas

O método `.describe()` traz contagem, média, desvio padrão, mínimo, máximo e os quartis (25%, 50%, 75%) de cada coluna numérica de uma só vez.


```python
df.describe()
```

Para uma coluna específica, dá para chamar cada estatística diretamente.


```python
# media e mediana
print(f"media:   {df.High.mean():.2f}")
print(f"mediana: {df.High.median():.2f}")
```

```python
# desvio padrao e variancia
print(f"std: {df.High.std():.2f}")
print(f"var: {df.High.var():.2f}")
```

```python
# minimo e maximo
print(f"min: {df.High.min():.2f}")
print(f"max: {df.High.max():.2f}")
```

### Mediana é mais resistente a outliers

Considere um conjunto pequeno com um valor extremo: `[1, 1, 2, 4, 900]`. A média é puxada pelo `900`. A mediana ignora esse efeito.


```python
amostra = pd.Series([1, 1, 2, 4, 900])
print(f"media:   {amostra.mean()}")
print(f"mediana: {amostra.median()}")
```

## 7. Criando DataFrames do zero

Nem sempre você parte de um `csv`. Em testes, simulações e exemplos didáticos, você vai querer construir um `DataFrame` na hora.

### A partir de um dicionário

A chave do dicionário vira o nome da coluna; o valor vira a coluna em si.


```python
dados = {
    "nome": ["Carlos", "Theo", "Raquel", "Fernanda"],
    "idade": [35, 32, 15, 49],
    "cidade": ["Araraquara", "Belém", "Natal", "Curitiba"],
    "comprou": [True, False, False, True],
}

clientes = pd.DataFrame(dados)
clientes
```

Posso definir o índice manualmente, por exemplo, usando o id de cada usuário.


```python
clientes.index = [8712, 5831, 4421, 9873]
clientes
```

### A partir de listas

Quando os dados estão em listas separadas, dá para passar tudo de uma vez como uma lista de listas.


```python
linhas = [
    ["Carlos", 35, "Araraquara", True],
    ["Theo", 32, "Belém", False],
    ["Raquel", 15, "Natal", False],
    ["Fernanda", 49, "Curitiba", True],
]

clientes2 = pd.DataFrame(
    linhas,
    columns=["nome", "idade", "cidade", "comprou"],
)
clientes2
```

### A partir de NumPy

Para simulações, é comum gerar dados aleatórios com `numpy` e empacotar num `DataFrame`.


```python
rng = np.random.default_rng(seed=42)

amostras = pd.DataFrame({
    "altura_cm":  rng.normal(loc=170, scale=10, size=200),
    "peso_kg":    rng.normal(loc=70,  scale=12, size=200),
    "horas_sono": rng.normal(loc=7,   scale=1.2, size=200),
})

amostras.describe()
```

## 8. Tratando valores ausentes

Dado real quase nunca vem completo. Para essa parte, vamos usar um dataset de **corridas de bicicleta compartilhada** (Brasília, 2018), com cerca de 287 mil registros e vários campos com nulos.


```python
url_rides = "https://dl.dropboxusercontent.com/s/yyfeoxqw61o3iel/df_rides.csv"
rides = pd.read_csv(url_rides, parse_dates=["user_birthdate", "ride_date"])
rides.shape
```

```python
rides.head()
```

### Detectando valores ausentes

`isna()` devolve um `DataFrame` booleano (`True` onde o valor é nulo). Combinado com `.sum()`, conta nulos por coluna.


```python
rides.isna().sum()
```

Mais de 60% dos registros não tem `user_residence` informada, e cerca de 43 mil corridas não tem `time_end` (provavelmente registros incompletos do app).


### Removendo linhas com nulos: `dropna`

`dropna()` apaga linhas que têm pelo menos um valor ausente. Use o argumento `subset` para olhar apenas determinadas colunas.


```python
# antes
print(f"antes:  {len(rides):,} linhas")

rides_limpo = rides.dropna(subset=["user_residence"])

print(f"depois: {len(rides_limpo):,} linhas")
```

### Preenchendo nulos: `fillna`

Em muitos casos, jogar fora a linha inteira é caro demais. Uma alternativa pragmática é preencher: mediana para colunas numéricas, moda (valor mais frequente) ou uma categoria explícita para categóricas.


```python
# preencher residencia ausente com a moda
moda_residencia = rides["user_residence"].mode()[0]
print(f"moda: {moda_residencia}")

rides["user_residence"] = rides["user_residence"].fillna(moda_residencia)
rides.isna().sum()
```

> **Cuidado**: imputar com mediana ou moda é uma escolha pragmática quando você não tem informação a priori. Em pipelines de machine learning, isso vira parte de um `Pipeline` do `scikit-learn`, e é fundamental calcular estatísticas só no conjunto de treino para evitar **data leakage**.
